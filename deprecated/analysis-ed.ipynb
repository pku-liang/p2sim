```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import struct
```

```python
with open('build/ed/rocket-m/coremark.bin', 'rb') as f:
    data = f.read()

N, = struct.unpack('q', data[:8])
act_cnt = np.frombuffer(data[8:8 + 4 * N], dtype=np.uint32)
act_hash = np.frombuffer(data[8 + 4 * N + 8: 8 + 4 * N + 8 + 2 * N * 32], dtype=np.uint16)
assert len(data) == 8 + 4 * N + 8 + 2 * N * 32
act_hash = act_hash.reshape(N, 32)
```

```python
plt.plot(act_cnt / act_cnt.max())
```

```python
sim = (act_hash[:, None, :] == act_hash[None, :, :]).sum(axis=-1)
```

```python
plt.imshow(sim)
```
